```python
n = int(input())
arr = list(map(int, input().split()))

bad = []

for i in range(n - 1):
    if arr[i] > arr[i + 1]:
        bad.append(i)

if len(bad) != 1:
    print(-1)
else:
    i = bad[0]

    # Can we change arr[i]?
    if (i == 0 or arr[i - 1] <= arr[i + 1]):
        print(i)

    # Can we change arr[i+1]?
    elif (i + 2 == n or arr[i] <= arr[i + 2]):
        print(i + 1)

    else:
        print(-1)
```

```python
n = int(input())
years = list(map(int, input().split()))
parties = input().split()

max_count = 1
count = 1
answer = None

for i in range(1, n):
    if parties[i] == parties[i - 1]:
        count += 1
    else:
        if count > max_count:
            max_count = count
            answer = years[i]
        count = 1

# Check if the longest streak is at the end
if count > max_count:
    max_count = count
    answer = None

print(answer)
```

```python
import calendar

month, year = input().split()

months = {
    "January": 1,
    "February": 2,
    "March": 3,
    "April": 4,
    "May": 5,
    "June": 6,
    "July": 7,
    "August": 8,
    "September": 9,
    "October": 10,
    "November": 11,
    "December": 12
}

days = calendar.monthrange(int(year), months[month])[1]

count = 0

for n in range(2, days + 1):
    factors = 0

    for i in range(1, n + 1):
        if n % i == 0:
            factors += 1

    if factors == 2:
        count += 1

print(count)
```
