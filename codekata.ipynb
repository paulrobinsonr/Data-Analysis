```python
N = int(input())
arr = list(map(int, input().split()))

sorted_arr = sorted(arr)

diff = []

for i in range(N):
    if arr[i] != sorted_arr[i]:
        diff.append(i)

if len(diff) == 1:
    print(diff[0])
else:
    print(-1)
```

```python
n = int(input())
arr = list(map(int, input().split()))

bad = []

for i in range(n - 1):
    if arr[i] > arr[i + 1]:
        bad.append(i)

if len(bad) != 1:
    print(-1)
else:
    i = bad[0]

    # Can we change arr[i]?
    if (i == 0 or arr[i - 1] <= arr[i + 1]):
        print(i)

    # Can we change arr[i+1]?
    elif (i + 2 == n or arr[i] <= arr[i + 2]):
        print(i + 1)

    else:
        print(-1)
```

```python
n,x=map(int,input().split())
arr=list(map(int,input().split()))
found=False
for i in range(n):
    for j in range(i+1,n):
        if arr[i] + arr[j] == x:
          found=True
          break
    if found:
        break
if found:
  print('yes')
else:
    print('no')
  
```

```python
n=int(input())
arr=list(map(int,input().split()))
r=[]
for i in arr:
    if i<n:
        r.append(i)
if len(r)==0:
    print(-1)
else:
    r.sort(reverse=True)
    print(*r)
```
