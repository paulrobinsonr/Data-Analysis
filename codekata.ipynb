```python
N = int(input())
arr = list(map(int, input().split()))

sorted_arr = sorted(arr)

diff = []

for i in range(N):
    if arr[i] != sorted_arr[i]:
        diff.append(i)

if len(diff) == 1:
    print(diff[0])
else:
    print(-1)
```

```python
n = int(input())
arr = list(map(int, input().split()))

bad = []

for i in range(n - 1):
    if arr[i] > arr[i + 1]:
        bad.append(i)

if len(bad) != 1:
    print(-1)
else:
    i = bad[0]

    # Can we change arr[i]?
    if (i == 0 or arr[i - 1] <= arr[i + 1]):
        print(i)

    # Can we change arr[i+1]?
    elif (i + 2 == n or arr[i] <= arr[i + 2]):
        print(i + 1)

    else:
        print(-1)
```

```python
n = int(input())

years = list(map(int, input().split()))
parties = input().split()

elections = list(zip(years, parties))
elections.sort()

count = 1

for i in range(1, n):
    if elections[i][1] == elections[i - 1][1]:
        count += 1
    else:
        if count >= 2:
            print(elections[i][0])
            break
        count = 1
```

```python
n = int(input())
years = list(map(int, input().split()))
parties = input().split()

max_count = 1
count = 1
answer = None

for i in range(1, n):
    if parties[i] == parties[i - 1]:
        count += 1
    else:
        if count > max_count:
            max_count = count
            answer = years[i]
        count = 1

# Check if the longest streak is at the end
if count > max_count:
    max_count = count
    answer = None

print(answer)
```
