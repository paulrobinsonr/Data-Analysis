```python
import pandas as pd

data = {
    "Student": [
        "Paul", "John", "Alex", "Mary", "David",
        "Sophia", "James", "Emma", "Daniel", "Olivia"
    ],
    "Math": [85, 72, 90, 88, 76, 95, 67, 82, 79, 91],
    "Science": [80, 78, 95, 84, 70, 96, 65, 88, 81, 92],
    "English": [78, 85, 89, 90, 75, 94, 72, 80, 77, 93],
    "City": [
        "Chennai", "Madurai", "Chennai", "Coimbatore", "Madurai",
        "Chennai", "Salem", "Coimbatore", "Madurai", "Chennai"
    ],
    "Gender": [
        "Male", "Male", "Male", "Female", "Male",
        "Female", "Male", "Female", "Male", "Female"
    ],
    "Attendance": [92, 85, 98, 96, 88, 99, 80, 94, 90, 97]
}

df = pd.DataFrame(data)

print(df)
```

```python
print(df.loc[df['Student']=='mary'])
```

```python
print(df.loc[df['Student']=='Mary','Math'])
```

```python
print(df.loc[df["Student"] == "Paul", ["Math", "Science"]])
```

```python
print(df.loc[df['Math'] > 90,['Student','Math']])
```

```python
print(df.loc[df['City']=='Chennai',['Student','Science','Math','English']])
```

```python
print(df.loc[(df['Gender']=='Male') & (df['Math']>80),['Student','City']])
```

```python
print(df[df['Gender']=='Male'].value_counts())
```

```python
print(df['Gender'].value_counts())
```

```python
print(df['Gender'].value_counts(normalize=True))
```

```python
print(df[df['Attendance']>90])
```

```python
print(df[df['Attendance']>90])
```

```python
print(df[(df['Gender']=='Female')& (df['Attendance']>=94)])
```
